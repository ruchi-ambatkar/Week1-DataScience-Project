```python
plt.figure(figsize=(7, 5))

sns.countplot(data=cleaned_df, x="Survived")

plt.title("Passenger Survival Distribution")
plt.xlabel("Survival Status")
plt.ylabel("Number of Passengers")

plt.xticks([0, 1], ["Did Not Survive", "Survived"])

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=cleaned_df, x="Survived")

plt.title("Passenger Survival Distribution")
plt.xlabel("Survival Status")
plt.ylabel("Number of Passengers")

plt.xticks([0, 1], ["Did Not Survive", "Survived"])

plt.tight_layout()

plt.savefig("../visualizations/survival_distribution.png")

plt.show()
```

### Interpretation

The survival distribution shows that more passengers did not survive than survived. Out of 891 passengers, 342 survived while 549 did not survive. The overall survival rate was 38.38%.

```python
cleaned_df["Sex"].value_counts()
```

```python
cleaned_df.groupby("Sex")["Survived"].mean() * 100
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=cleaned_df, x="Sex", hue="Survived")

plt.title("Passenger Survival by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")

plt.legend(
    title="Survival",
    labels=["Did Not Survive", "Survived"]
)

plt.tight_layout()

plt.savefig("../visualizations/survival_by_gender.png")

plt.show()
```

### Interpretation

The visualization shows a difference in survival outcomes between male and female passengers. The calculated survival rate was 74.20% for female passengers and 18.89% for male passengers in this dataset.

```python
cleaned_df["Pclass"].value_counts().sort_index()
```

```python
cleaned_df.groupby("Pclass")["Survived"].mean() * 100
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(data=cleaned_df, x="Age", bins=30, kde=True)

plt.title("Age Distribution of Passengers")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")

plt.tight_layout()

plt.savefig("../visualizations/age_distribution.png")

plt.show()
```

### Interpretation

The age distribution shows that passengers belonged to a wide range of age groups. The average passenger age was 29.36 years, while the median age was 28 years. The minimum recorded age was 0.42 years and the maximum was 80 years.

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=cleaned_df, x="Pclass", hue="Survived")

plt.title("Passenger Survival by Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")

plt.legend(
    title="Survival",
    labels=["Did Not Survive", "Survived"]
)

plt.tight_layout()

plt.savefig("../visualizations/survival_by_class.png")

plt.show()
```

### Interpretation

The visualization shows differences in survival rates across passenger classes. The calculated survival rates were 62.96% for first-class passengers, 47.28% for second-class passengers, and 24.24% for third-class passengers.

```python
plt.figure(figsize=(8, 6))

correlation = cleaned_df.select_dtypes(include="number").corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.tight_layout()

plt.savefig("../visualizations/correlation_heatmap.png")

plt.show()
```

### Interpretation

The correlation heatmap shows the relationships between the numerical variables in the dataset. It provides a visual overview of how variables such as survival, passenger class, age, fare, number of siblings or spouses, and number of parents or children are related. The heatmap can help identify relationships that may be explored further.
